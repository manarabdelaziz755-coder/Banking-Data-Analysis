```python
top_10_customer = transaction.head(10)
top_10_customer
```

```python
customer=customer.groupby("Branch_ID")["Customer_ID"].count()
customer = customer.reset_index()
customer.rename(columns={"Customer_ID":"Customer_Count"},inplace=True)
```

```python
average_balance = transaction["Total_Balance"].mean()
average_balance
high_balance_customers=transaction[transaction["Total_Balance"]>average_balance]
high_balance_customers.head()
high_balance_customers.shape[0]
```

```python
limit = 3000
high_value_transactions = transaction[transaction["Transaction_Amount"]>limit]
high_value_transactions.head()
high_value_transactions.shape[0]
```

```python
high_value_transactions["Transaction_Amount"].sum()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(high_value_transactions["Transaction_Amount"],bins=30)
plt.title("High Value Transactions Distribution")
plt.show()
```

```python
customer_transaction = customer.merge(
    transaction,
    on="Customer_ID",
    how="inner"
)
```

```python
avg_blaance = customer_transaction["Total_Balance"].mean()
senior_high_balance=customer_transaction[(customer_transaction["Age"]>50) & (customer_transaction["Total_Balance"]>avg_blaance)]
senior_high_balance.head()
senior_high_balance.shape[0]
```

```python
customer_transaction_count =transaction.groupby("Customer_ID")["Transaction_ID"].count()
customer_transaction_count = customer_transaction_count.reset_index()
customer_transaction_count.rename(columns={"Transaction_ID":"Transaction_count"},inplace=True)
avg_transactions = customer_transaction_count["Transaction_count"].mean()
avg_transactions

high_activity_customers = customer_transaction_count[
    customer_transaction_count["Transaction_count"] > avg_transactions
]

high_activity_customers.sort_values(
    by="Transaction_count",
    ascending=False
).head(10)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=top_10_customer,
            x="Customer_ID",y="Transaction_Amount")
plt.title("Top 10 Customers by Transaction Amount")
plt.xlabel("Customer_ID")
plt.ylabel("Transaction_Amount")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(transaction['Transaction_Amount'], bins=30)
plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
investment_counts = transaction["Investment_Type"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    investment_counts,
    labels=investment_counts.index,
    autopct="%1.1f%%"
)

plt.title("Investment Type Distribution")

plt.show()
```

```python
transaction["Transaction_Date"] = pd.to_datetime(
    transaction["Transaction_Date"]
)

transaction["Month"] = transaction[
    "Transaction_Date"
].dt.month

monthly_amount = transaction.groupby(
    "Month"
)["Transaction_Amount"].sum().reset_index()
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(data=monthly_amount,x="Month",y="Transaction_Amount")

plt.title("Monthly Transaction Trend")
plt.xlabel("Month")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
avg_investment = transaction.groupby(
    "Investment_Type"
)["Investment_Amount"].mean().reset_index()

plt.figure(figsize=(10,5))

sns.barplot(
    data=avg_investment,
    x="Investment_Type",
    y="Investment_Amount"
)

plt.title("Average Investment Amount by Type")

plt.xticks(rotation=45)

plt.show()
```
